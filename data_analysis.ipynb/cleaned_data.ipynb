```python
import pandas as pd

# 1. LOAD DATA
df = pd.read_csv('ai4i2020.csv')

# 2. RENAME COLUMNS (Standardizing for SQL/Tableau)
df.columns = ['UDI', 'Product_ID', 'Type', 'Air_Temp_K', 'Process_Temp_K',
              'Rotational_Speed_RPM', 'Torque_Nm', 'Tool_Wear_Min',
              'Machine_Failure', 'TWF', 'HDF', 'PWF', 'OSF', 'RNF']

# 3. FEATURE ENGINEERING (The "Industrial Engineering" Touch)
# Temp_Delta: High efficiency usually means better heat management
df['Temp_Delta'] = df['Process_Temp_K'] - df['Air_Temp_K']

# Power_Watts: Physical work done (Torque * Speed * Constant)
# (Constant 2*pi/60 is standard, but we'll use raw product for relative comparison)
df['Power_Watts'] = df['Torque_Nm'] * df['Rotational_Speed_RPM']

# 4. CRITICAL INSIGHTS GENERATION (Save these numbers for your README!)
print("--- README DATA GENERATOR ---")

# Insight A: Failure Rate by Type
type_stats = df.groupby('Type')['Machine_Failure'].agg(['count', 'sum'])
type_stats['Failure_Rate'] = (type_stats['sum'] / type_stats['count']) * 100
print("\n1. Failure Rates by Product Type:")
print(type_stats)

# Insight B: Temperature & Failures
avg_delta_fail = df[df['HDF'] == 1]['Temp_Delta'].mean()
avg_delta_ok = df[df['HDF'] == 0]['Temp_Delta'].mean()
print(f"\n2. Avg Temp Delta during Heat Failure: {avg_delta_fail:.2f} K")
print(f"   Avg Temp Delta during Normal Ops: {avg_delta_ok:.2f} K")

# Insight C: Total Downtime Events
total_failures = df['Machine_Failure'].sum()
print(f"\n3. Total Machine Failures (Downtime Events): {total_failures}")

# 5. EXPORT FOR TABLEAU
df.to_csv('smart_factory_cleaned.csv', index=False)
print("\n--- Success! 'smart_factory_cleaned.csv' saved. ---")
```
